```python
#Dropout is negatively associate with effectiveness
df["retention_rate"] = 1 - df["dropout_rate"]

df[["dropout_rate","retention_rate"]].head()
```

```python
#Handling Missing Values
numeric_features = [
    "completion_rate",
    "retention_rate",
    "avg_score_improvement",
    "avg_quiz_score",
    "avg_watch_time",
    "assignment_submission_rate",
    "forum_activity_rate",
    "avg_feedback_score",
    "feedback_response_rate"
]

from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
df[numeric_features] = imputer.fit_transform(df[numeric_features])
```

```python
df[numeric_features].isnull().sum()
```

```python
#aggregrate the batch data to instructor level
instructor_df = df.groupby("instructor_id").agg(
    completion_mean=("completion_rate", "mean"),
    completion_std=("completion_rate", "std"),

    retention_mean=("retention_rate", "mean"),

    score_improvement_mean=("avg_score_improvement", "mean"),
    score_improvement_std=("avg_score_improvement", "std"),

    quiz_mean=("avg_quiz_score", "mean"),

    watch_time_mean=("avg_watch_time", "mean"),

    assignment_mean=("assignment_submission_rate", "mean"),

    forum_mean=("forum_activity_rate", "mean"),

    feedback_mean=("avg_feedback_score", "mean"),

    feedback_response_mean=("feedback_response_rate", "mean"),

    batch_count=("batch_id", "count")
).reset_index()

instructor_df.head()
```

```python
std_columns = ["completion_std","score_improvement_std"]

for column in std_columns:
    instructor_df[column] = instructor_df[column].fillna(0)

print("Instructor-level dataset shape:", instructor_df.shape)
```

```python
instructor_df.head(10)
```

```python
from sklearn.preprocessing import MinMaxScaler

effectiveness_features = [
    "completion_mean",
    "retention_mean",
    "score_improvement_mean",
    "quiz_mean",
    "watch_time_mean",
    "assignment_mean",
    "forum_mean",
    "feedback_mean",
    "feedback_response_mean"
]

scaler = MinMaxScaler()

instructor_df[effectiveness_features] = scaler.fit_transform(
    instructor_df[effectiveness_features]
)

instructor_df[effectiveness_features].head()
```

```python
instructor_df["effectiveness_score"] = (
    0.15 * instructor_df["completion_mean"] +
    0.10 * instructor_df["retention_mean"] +
    0.15 * instructor_df["score_improvement_mean"] +
    0.10 * instructor_df["quiz_mean"] +
    0.10 * instructor_df["watch_time_mean"] +
    0.10 * instructor_df["assignment_mean"] +
    0.10 * instructor_df["forum_mean"] +
    0.10 * instructor_df["feedback_mean"] +
    0.10 * instructor_df["feedback_response_mean"]
)

instructor_df[
    ["instructor_id", "effectiveness_score"]
].head(10)
```

```python
instructor_df["effectiveness_tier"] = pd.qcut(instructor_df["effectiveness_score"],q=3,labels=["Low", "Medium", "High"])

instructor_df[
    ["instructor_id", "effectiveness_score", "effectiveness_tier"]
].head(10)
```

```python
tier_counts = instructor_df["effectiveness_tier"].value_counts()

plt.figure(figsize=(7, 5))

sns.countplot(data=instructor_df,x="effectiveness_tier")

plt.title("Instructor Effectiveness Tier Distribution")
plt.xlabel("Effectiveness Tier")
plt.ylabel("Number of Instructors")
```

```python
model_features = [
    "completion_mean",
    "retention_mean",
    "score_improvement_mean",
    "quiz_mean",
    "watch_time_mean",
    "assignment_mean",
    "forum_mean",
    "feedback_mean",
    "feedback_response_mean",
    "batch_count"
]

X = instructor_df[model_features]

y = instructor_df["effectiveness_tier"]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

#Logistic Regression Model is used
logistic_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", MinMaxScaler()),
    ("model", LogisticRegression(max_iter=1000,class_weight="balanced", random_state=42))])

logistic_model.fit(X_train, y_train)

logistic_pred = logistic_model.predict(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier

#random forest is used
rf_model = Pipeline([
    ("imputer",SimpleImputer(strategy="median")),
    ("model",RandomForestClassifier(n_estimators=1000,random_state=42,class_weight="balanced"))
])

rf_model.fit(X_train,y_train)
rf_pred = rf_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score


def evaluate_model(model_name, y_true, y_pred):
    
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(
        y_true, y_pred,
        average="macro",
        zero_division=0
    )
    recall = recall_score(
        y_true, y_pred,
        average="macro",
        zero_division=0
    )
    f1 = f1_score(
        y_true, y_pred,
        average="macro",
        zero_division=0
    )
    
    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Macro Precision": precision,
        "Macro Recall": recall,
        "Macro F1": f1
    }
```

```python
results = pd.DataFrame([
    evaluate_model(
        "Logistic Regression",
        y_test,
        logistic_pred
    ),
    
    evaluate_model(
        "Random Forest",
        y_test,
        rf_pred
    )
])

results
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(
    y_test,
    rf_pred,
    labels=["Low", "Medium", "High"]
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Low", "Medium", "High"],
    yticklabels=["Low", "Medium", "High"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Random Forest Confusion Matrix")
```

```python
from sklearn.model_selection import StratifiedKFold,cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = cross_val_score(
    rf_model,
    X,
    y,
    cv=cv,
    scoring="f1_macro"
)

print("Cross-validation F1 scores:")
print(cv_scores)

print("\nMean Macro F1:", cv_scores.mean())
print("Standard Deviation:", cv_scores.std())
```

```python
rf_classifier = rf_model.named_steps["model"]

feature_importance = pd.Series(
    rf_classifier.feature_importances_,
    index=model_features
).sort_values(ascending=False)

feature_importance
```

## 1. Which features most influenced instructor effectiveness, and why?

The most influential features were identified using the Random Forest feature importance values.

Learning outcome features such as completion rate, score improvement, and quiz performance are expected to be important because they directly represent learner outcomes. Engagement variables such as assignment submission and watch time provide additional information about learner participation.

However, feature importance represents predictive association rather than causation. A feature being important does not prove that changing that feature will directly improve instructor effectiveness.

## 2. Which variables could be misleading or confounded?

Several variables could be affected by factors other than instructor quality.

- Quiz scores may depend on learners' prior knowledge.
- Completion rate may depend on course difficulty, learner motivation, or course duration.
- Watch time may be misleading because longer watch time could mean either strong engagement or difficulty understanding the material.
- Feedback scores may be influenced by instructor personality, grading strictness, or learner expectations.
- Forum activity may depend on the nature of the course rather than instructor effectiveness.

Course difficulty, learner background, batch composition, and assessment design could therefore confound the relationship between instructor characteristics and learner outcomes.

## 3. How could this model fail in real-world usage?

The model could fail in several situations.

1. **New instructors:** An instructor with only one or two batches may not have enough historical evidence for a reliable estimate.

2. **Different course difficulty:** Instructors teaching more difficult courses may naturally have lower completion or quiz scores.

3. **Different learner populations:** Some batches may contain highly motivated or experienced learners, while others may require much more support.

4. **Feedback bias:** Learner feedback may reflect personality or expectations rather than actual teaching effectiveness.

5. **Distribution shift:** Future learners, courses, or teaching methods may differ from the data used to train the model.

6. **Small sample size:** A limited number of instructors can make model estimates unstable.

Therefore, the model should be treated as an analytical support tool rather than an unquestionable decision-making system.

## 4. What additional data would you want to improve this analysis?

Additional useful data would include:

### Learner-level data
- Pre-course assessment score
- Prior knowledge
- Attendance
- Learning time
- Learner demographics where appropriate

### Course-level data
- Course difficulty
- Course duration
- Course category
- Assessment difficulty
- Course format

### Instructor-level data
- Years of teaching experience
- Previous experience with the course
- Certifications
- Subject expertise

### Batch-level data
- Batch size
- Learner composition
- Online/offline format
- Class schedule

### Long-term outcomes
- Post-course assessment
- Knowledge retention
- Certification completion
- Job placement or career outcomes where relevant

These additional variables would help separate instructor effects from learner and course effects.

## 5. Should this model be used for instructor performance evaluation?

The model should not be used as the sole basis for instructor performance evaluation.

Instructor effectiveness is influenced by many factors that may not be captured in the available data, including course difficulty, learner background, batch composition, and instructor experience.

Instead, the model should be used as a decision-support tool. For example, if an instructor consistently receives a low effectiveness score, the result could trigger a deeper review of learner feedback, course difficulty, assessment design, and teaching practices.

Human review should remain part of the final decision, especially when the results could affect promotions, compensation, or employment.
